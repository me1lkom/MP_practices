```python

from sympy import *
from IPython.display import display, Math

init_printing()

# Задание 2.1

# а)
z_a = (5 + I) * (7 - 6*I) / (3 + I)

# б)
z_b = (1 + I)**5 / (1 - I)**3

# в)
z_c = (
    (-1 + I*sqrt(3))**15 / (1 - I)**20
    + (-1 - I*sqrt(3))**15 / (1 + I)**20
)

for name, expression in [
    ("а", z_a),
    ("б", z_b),
    ("в", z_c)
]:
    z = simplify(expand_complex(expression))

    # Модуль
    r = simplify(Abs(z))

    # Аргумент
    phi = arg(z)

    print(f"\nЗадание {name}")

    print("Алгебраическая форма:")
    display(Math(rf"z={latex(z)}"))

    print("Тригонометрическая форма:")
    display(Math(
        rf"z={latex(r)}\left(\cos({latex(phi)})"
        rf"+i\sin({latex(phi)})\right)"
    ))

    print("Показательная форма:")
    display(Math(
        rf"z={latex(r)}e^{{i({latex(phi)})}}"
    ))
```

```python

# 2.2

import numpy as np
import matplotlib.pyplot as plt

n = 8
r = np.sqrt(2)
phi = np.pi / 4

roots = []

for k in range(n):
    angle = (phi + 2*np.pi*k) / n

    z = r**(1/n) * np.exp(1j*angle)
    roots.append(z)

    print(f"z{k} = {z.real:.6f} {z.imag:+.6f}i")

roots = np.array(roots)

fig, ax = plt.subplots(figsize=(7, 7))

ax.scatter(roots.real, roots.imag, color="red")

for k, z in enumerate(roots):
    ax.annotate(
        f"z{k}",
        (z.real, z.imag),
        xytext=(5, 5),
        textcoords="offset points"
    )

t = np.linspace(0, 2*np.pi, 500)
radius = r**(1/n)

ax.plot(
    radius*np.cos(t),
    radius*np.sin(t),
    "--"
)

ax.axhline(0, color="black")
ax.axvline(0, color="black")
ax.set_aspect("equal")
ax.set_xlabel("Re(z)")
ax.set_ylabel("Im(z)")
ax.set_title("z^8 = 1+i")
ax.grid(True)

plt.show()
```

```python

# 2.3

import numpy as np
import matplotlib.pyplot as plt

def complex_roots(w, n):
    r = abs(w)
    phi = np.angle(w)

    return np.array([
        r**(1/n) * np.exp(1j*(phi + 2*np.pi*k)/n)
        for k in range(n)
    ])

examples = {
    "a": (1, 8),
    "b": (-18/(1+1j*np.sqrt(3)), 4),
    "c": (
        (1-5j)/(1+1j)
        - 5*(1+2j)/(2-1j) + 11,
        3
    ),
    "d": (1-1j, 9)
}

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

for ax, (name, (w, n)) in zip(axes.flat, examples.items()):

    roots = complex_roots(w, n)

    print(f"\n{name})")
    for k, z in enumerate(roots):
        print(f"z{k} = {z.real:.6f} {z.imag:+.6f}i")

    ax.scatter(roots.real, roots.imag)

    for k, z in enumerate(roots):
        ax.annotate(
            f"z{k}",
            (z.real, z.imag),
            xytext=(5, 5),
            textcoords="offset points"
        )

    ax.axhline(0, color="black")
    ax.axvline(0, color="black")
    ax.set_aspect("equal")
    ax.grid(True)
    ax.set_title(f"2.3 ({name})")
    ax.set_xlabel("Re(z)")
    ax.set_ylabel("Im(z)")

plt.tight_layout()
plt.show()
```

```python

# 2.5

import sympy as sp
from IPython.display import display

A = sp.symbols("A", positive=True, real=True)
C, p, q = sp.symbols("C p q", real=True)
x, y = sp.symbols("x y", real=True)

i = sp.I

B = p + i*q
z = x + i*y

# Исходное уравнение
equation = sp.expand(
    A*(x**2 + y**2)
    + sp.conjugate(B)*z
    + B*sp.conjugate(z)
    + C
)

# Центр окружности
center = -B/A

# Радиус
radius = sp.sqrt(p**2 + q**2 - A*C)/A

print("Исходное уравнение:")
display(sp.Eq(equation, 0))

print("Уравнение окружности:")
display(
    sp.Eq(
        (x + p/A)**2 + (y + q/A)**2,
        (p**2 + q**2 - A*C)/A**2
    )
)

print("Центр окружности:")
display(center)

print("Радиус окружности:")
display(radius)
```

```python

# 2.6

import sympy as sp

x = sp.symbols("x")

p = (
    x**11 + x**10 + x**8
    - 2*x**7 - x**5 + x**4
    + 2*x**2 + 2*x + 2
)

# Находим все 11 комплексных корней
roots = sp.nroots(p, n=18, maxsteps=300)

print("Все корни многочлена:")

for k, z in enumerate(roots, start=1):
    print(f"z{k} = {z}")

# Сумма модулей
sum_abs = sum(abs(complex(z)) for z in roots)

# Максимальная действительная часть
max_real = max(complex(z).real for z in roots)

print("\nСумма модулей:", round(sum_abs, 6))
print("Максимальная действительная часть:", round(max_real, 6))
```

```python

# 2.7

import numpy as np
import mpmath as mp
import matplotlib.pyplot as plt
from scipy.special import loggamma

mp.mp.dps = 30

# a) C_(1+i)^(1-i)
z = 1 + 1j
w = 1 - 1j

C = mp.gamma(z+1) / (
    mp.gamma(w+1) * mp.gamma(z-w+1)
)

print("a) Биномиальный коэффициент:")
print(mp.nstr(C, 12))

# б) Аргумент
angle = mp.arg(C)

print("\nб) Аргумент:")
print("В радианах:", float(angle))
print("В градусах:", float(mp.degrees(angle)))

# в) Линии уровня Arg(C_z^(1/z))

x = np.linspace(-3.2, 3.2, 500)
y = np.linspace(-3.2, 3.2, 500)

X, Y = np.meshgrid(x, y)
Z = X + 1j*Y

with np.errstate(all="ignore"):
    W = 1/Z

    logC = (
        loggamma(Z+1)
        - loggamma(W+1)
        - loggamma(Z-W+1)
    )

    phase = np.angle(np.exp(1j*np.imag(logC)))

phase[~np.isfinite(phase)] = np.nan
phase[np.abs(Z) < 0.04] = np.nan

fig, ax = plt.subplots(figsize=(8, 7))

levels = np.linspace(-np.pi, np.pi, 24)

plot = ax.contourf(
    X, Y, phase,
    levels=levels,
    cmap="twilight"
)

ax.contour(
    X, Y, phase,
    levels=levels[::2],
    colors="black",
    linewidths=0.35
)

plt.colorbar(plot, ax=ax)

ax.set_title("2.7: Линии уровня аргумента")
ax.set_xlabel("Re(z)")
ax.set_ylabel("Im(z)")
ax.set_aspect("equal")

plt.show()
```

```python

# 2.9 (в)

import numpy as np
import matplotlib.pyplot as plt

def julia(c, size=500, iterations=150):

    x = np.linspace(-1.8, 1.8, size)
    y = np.linspace(-1.8, 1.8, size)

    X, Y = np.meshgrid(x, y)

    z = X + 1j*Y

    result = np.zeros(z.shape)
    active = np.ones(z.shape, dtype=bool)

    for k in range(iterations):

        z[active] = z[active]**2 + c

        result[active] = k

        active &= (np.abs(z) <= 2)

        if not np.any(active):
            break

    return result


# Значения c из задания
values = [
    -0.1 + 0.65j,
    0.28 + 0.009j
]

fig, axes = plt.subplots(1, 2, figsize=(12, 6))

for ax, c in zip(axes, values):

    image = julia(c)

    ax.imshow(
        image,
        extent=[-1.8, 1.8, -1.8, 1.8],
        origin="lower",
        cmap="magma"
    )

    ax.set_title(f"c = {c}")
    ax.set_aspect("equal")

plt.tight_layout()
plt.show()
```

```python

# 2.9 (г)

import numpy as np
import matplotlib.pyplot as plt

# Генерируем приближение множеств Жюлиа
# методом случайных обратных итераций

def julia_inverse(which, walkers=10000, steps=75):

    rng = np.random.default_rng(42)

    z = (
        rng.uniform(-1, 1, walkers)
        + 1j*rng.uniform(-1, 1, walkers)
    )

    points = []

    for step in range(steps):

        with np.errstate(all="ignore"):

            if which == 1:
                # f(z) = (i*z^5 - 1)/(z^5 - i)
                w = (1j*z - 1)/(z - 1j)

                branch = rng.integers(0, 5, walkers)

                z = (
                    np.abs(w)**(1/5)
                    * np.exp(
                        1j*(np.angle(w) + 2*np.pi*branch)/5
                    )
                )

            else:
                # f(z) = (z^2 + i)/(1 - z^2)
                w = (z - 1j)/(z + 1)

                sign = rng.choice([-1, 1], size=walkers)

                z = sign*np.sqrt(w)

        if step > 20 and step % 2 == 0:
            valid = z[
                np.isfinite(z) & (np.abs(z) < 4)
            ]
            points.append(valid.copy())

    return np.concatenate(points)


fig, axes = plt.subplots(1, 2, figsize=(12, 6))

titles = [
    "f(z) = (i*z^5-1)/(z^5-i)",
    "f(z) = (z^2+i)/(1-z^2)"
]

for ax, number, title in zip(axes, [1, 2], titles):

    points = julia_inverse(number)

    ax.scatter(
        points.real,
        points.imag,
        s=0.1,
        color="darkblue",
        alpha=0.3
    )

    ax.set_title(title)
    ax.set_aspect("equal")
    ax.set_xlim(-2.3, 2.3)
    ax.set_ylim(-2.3, 2.3)

plt.tight_layout()
plt.show()
```

```python

# 2.10 — прямая, окружность, парабола, гипербола

import numpy as np
import matplotlib.pyplot as plt

def riemann(z):

    x = z.real
    y = z.imag

    d = 1 + x**2 + y**2

    X = 2*x/d
    Y = 2*y/d
    Z = (x**2 + y**2 - 1)/d

    return X, Y, Z


t = np.linspace(-3, 3, 1000)
angle = np.linspace(0, 2*np.pi, 1000)

curves = {
    "Прямая": [t + 0j],
    "Окружность": [np.exp(1j*angle)],
    "Парабола": [t + 1j*t**2],
    "Гипербола": [
        np.cosh(t) + 1j*np.sinh(t),
        -np.cosh(t) + 1j*np.sinh(t)
    ]
}

u = np.linspace(0, 2*np.pi, 40)
v = np.linspace(0, np.pi, 30)

SX = np.outer(np.cos(u), np.sin(v))
SY = np.outer(np.sin(u), np.sin(v))
SZ = np.outer(np.ones_like(u), np.cos(v))

fig = plt.figure(figsize=(12, 10))

for k, (name, parts) in enumerate(curves.items(), 1):

    ax = fig.add_subplot(2, 2, k, projection="3d")

    ax.plot_wireframe(
        SX, SY, SZ,
        color="gray",
        alpha=0.4,
        linewidth=0.3
    )

    for z in parts:
        X, Y, Z = riemann(z)
        ax.plot(X, Y, Z, color="red", linewidth=2)

    ax.set_title(name)
    ax.set_box_aspect([1, 1, 1])
    ax.set_xlim(-1.1, 1.1)
    ax.set_ylim(-1.1, 1.1)
    ax.set_zlim(-1.1, 1.1)

plt.tight_layout()
plt.show()
```

```python

# 2.10 — логарифмическая спираль на сфере Римана

import numpy as np
import matplotlib.pyplot as plt

# Параметры спирали (пример)
a = 0.3
b = 0.16

phi = np.linspace(-5*np.pi, 5*np.pi, 2500)

# Полярное уравнение
r = a*np.exp(b*phi)

# Комплексные координаты
z = r*np.exp(1j*phi)

# Обратная стереографическая проекция
x = z.real
y = z.imag

d = 1 + x*x + y*y

X = 2*x/d
Y = 2*y/d
Z = (x*x + y*y - 1)/d

# Поверхность сферы
u = np.linspace(0, 2*np.pi, 50)
v = np.linspace(0, np.pi, 35)

SX = np.outer(np.cos(u), np.sin(v))
SY = np.outer(np.sin(u), np.sin(v))
SZ = np.outer(np.ones_like(u), np.cos(v))

fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection="3d")

ax.plot_wireframe(
    SX, SY, SZ,
    color="gray",
    alpha=0.35,
    linewidth=0.3
)

ax.plot(X, Y, Z, color="red", linewidth=2)

ax.set_title("Логарифмическая спираль на сфере Римана")
ax.set_box_aspect([1, 1, 1])

plt.show()
```
